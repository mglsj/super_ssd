## 1. What is a Dictionary?

A dictionary is a Python data structure that stores data as **key-value pairs**. Unlike lists that use numerical indexes, dictionaries use meaningful keys to organize and access data.

Think of a dictionary like a real-world dictionary:
- **Key** = The word you're looking up
- **Value** = The definition of that word

## 2. Creating Dictionaries

### Basic Dictionary Creation

```python
# Empty dictionary
empty_dict = {}
empty_dict_alt = dict()

# Dictionary with initial values
student = {
    "name": "Alice",
    "age": 20,
    "grade": "A",
    "courses": ["Math", "Science"]
}

# Using dict() constructor
teacher = dict(name="Mr. Smith", subject="Math", years_experience=5)

print(student)
print(teacher)
```

### Different Key Types

```python
# Keys can be various immutable types
mixed_keys = {
    "string_key": "value1",      # String key
    123: "value2",               # Integer key
    (1, 2): "value3",            # Tuple key (immutable)
    True: "value4"               # Boolean key
}

# Values can be ANY data type
mixed_values = {
    "name": "Alice",             # String value
    "age": 25,                   # Integer value
    "grades": [85, 92, 78],      # List value
    "info": {"city": "NYC", "zip": 10001},  # Dictionary value
    "is_active": True            # Boolean value
}
```

## 3. Accessing Dictionary Values

### Using Square Brackets []

```python
student = {
    "name": "Bob",
    "age": 22,
    "major": "Computer Science",
    "gpa": 3.8
}

# Access values using keys
student_name = student["name"]
student_age = student["age"]
student_gpa = student["gpa"]

print(f"Name: {student_name}")    # Name: Bob
print(f"Age: {student_age}")      # Age: 22
print(f"GPA: {student_gpa}")      # GPA: 3.8
```

### Using get() Method (Safe Access)

```python
# get() returns None if key doesn't exist
phone = student.get("phone_number")
print(f"Phone: {phone}")  # Phone: None

# get() with default value
phone = student.get("phone_number", "Not provided")
print(f"Phone: {phone}")  # Phone: Not provided

# Check if key exists
if "major" in student:
    print(f"Major: {student['major']}")  # Major: Computer Science
```

## 4. Updating and Modifying Dictionaries

### Adding New Key-Value Pairs

```python
student = {"name": "Alice", "age": 20}

# Add new key-value pairs
student["major"] = "Biology"
student["gpa"] = 3.7

print(student)  # {'name': 'Alice', 'age': 20, 'major': 'Biology', 'gpa': 3.7}
```

### Modifying Existing Values

```python
# Update existing values
student["age"] = 21
student["gpa"] = 3.8

print(student)  # {'name': 'Alice', 'age': 21, 'major': 'Biology', 'gpa': 3.8}
```

### Multiple Updates with update()

```python
# Update multiple values at once
student.update({
    "age": 22,
    "minor": "Chemistry",
    "graduation_year": 2024
})

print(student)
# {'name': 'Alice', 'age': 22, 'major': 'Biology', 'gpa': 3.8, 
#  'minor': 'Chemistry', 'graduation_year': 2024}
```

### Nested dictionary updates (careful!)

```python
student = {
    "name": "Alice", 
    "grades": {"math": 85, "science": 90}
}

# This COMPLETELY replaces 'grades'
student.update({"grades": {"english": 88}})
print(student)
# {'name': 'Alice', 'grades': {'english': 88}}  # Lost math and science!

# To update nested dict, you need to update the inner dict
student = {
    "name": "Alice", 
    "grades": {"math": 85, "science": 90}
}
student["grades"].update({"english": 88, "history": 92})
print(student)
# {'name': 'Alice', 'grades': {'math': 85, 'science': 90, 'english': 88, 'history': 92}}
```

## 5. Removing Items from Dictionaries

```python
grades = {"math": 85, "science": 92, "history": 78, "english": 88}

# Remove specific key-value pair
removed_grade = grades.pop("history")
print(f"Removed: {removed_grade}")  # Removed: 78
print(f"Remaining: {grades}")       # {'math': 85, 'science': 92, 'english': 88}

# Clear all items
grades.clear()
print(f"After clear: {grades}")     # After clear: {}

# Delete entire dictionary
del grades
# print(grades)  # This would cause NameError
```

## 6. Dictionary Methods and Operations

### Common Dictionary Methods

```python
inventory = {"apples": 10, "bananas": 5, "oranges": 8}

# Get all keys
print(f"Keys: {list(inventory.keys())}")      # ['apples', 'bananas', 'oranges']

# Get all values  
print(f"Values: {list(inventory.values())}")  # [10, 5, 8]

# Get all key-value pairs
print(f"Items: {list(inventory.items())}")    # [('apples', 10), ('bananas', 5), ('oranges', 8)]

# Length of dictionary
print(f"Number of items: {len(inventory)}")   # Number of items: 3
```

### Iterating Through Dictionaries

```python
student_scores = {"Alice": 85, "Bob": 92, "Charlie": 78, "Diana": 96}

# Iterate through keys
print("Students:")
for student in student_scores:
    print(f"  - {student}")

# Iterate through key-value pairs
print("\nScores:")
for student, score in student_scores.items():
    print(f"  {student}: {score}")

# Iterate through values
for score in student_scores.values():
    print(f"Score: {score}")
```

## 7. Why Dictionaries Are Better Than Lists in Some Cases

### Case 1: Looking Up Information by Meaningful Keys

**With Lists (Less Efficient):**

```python
# Store student data in parallel lists
student_names = ["Alice", "Bob", "Charlie"]
student_ages = [20, 21, 19]
student_grades = [85, 92, 78]

# To find Bob's grade:
bob_index = student_names.index("Bob")  # Have to search through list
bob_grade = student_grades[bob_index]
print(f"Bob's grade: {bob_grade}")  # 92
```

**With Dictionaries (More Efficient):**

```python
# Store student data in dictionary
students = {
    "Alice": {"age": 20, "grade": 85},
    "Bob": {"age": 21, "grade": 92},
    "Charlie": {"age": 19, "grade": 78}
}

# Direct access - no searching needed
bob_grade = students["Bob"]["grade"]
print(f"Bob's grade: {bob_grade}")  # 92
```

### Case 2: Structured Data Representation

**With Lists (Hard to Understand):**

```python
# Student record as list
student1 = ["Alice", 20, "Computer Science", 3.8, "New York"]
# What is index 2? index 4?
major = student1[2]  # Not clear what this represents
```

**With Dictionaries (Self-Documenting):**

```python
# Student record as dictionary
student1 = {
    "name": "Alice",
    "age": 20,
    "major": "Computer Science", 
    "gpa": 3.8,
    "city": "New York"
}

major = student1["major"]  # Clear what we're accessing
```

## Key Takeaways

- **Dictionaries store key-value pairs** for organized data
- **Keys must be immutable** (strings, numbers, tuples)
- **Values can be any data type**
- **Use square brackets []** for access (risks KeyError)
- **Use get() method** for safe access with defaults
- **Dictionaries are faster for lookups** than searching through lists
- **Use dictionaries when you need meaningful labels** for your data
- **Dictionaries make code more readable** and self-documenting

**Remember:** Choose dictionaries when you need to look up data by meaningful keys rather than numerical positions!
